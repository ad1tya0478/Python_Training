```python
# Matplotlib is a python library used to create graphs and visualizations from data.


import matplotlib.pyplot as plt
import numpy as np

x = np.array([2023, 2024, 2025, 2026])
y = np.array([ 15, 25, 30, 20])

# x -> position/input  || y -> value/output
plt.plot(x, y)
```
